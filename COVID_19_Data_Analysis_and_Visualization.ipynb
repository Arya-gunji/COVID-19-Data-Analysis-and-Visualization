```python
import matplotlib.pyplot as plt
import seaborn as sns

corr = data[[
    "Confirmed",
    "Deaths",
    "Recovered",
    "Active"
]].corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("COVID-19 Dataset Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=data,
    x="Confirmed",
    y="Deaths",
    hue="WHO Region",
    s=80
)

plt.title("COVID-19: Confirmed Cases vs Deaths")
plt.xlabel("Confirmed Cases")
plt.ylabel("Deaths")
plt.tight_layout()
plt.show()
```
